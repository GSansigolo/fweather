```python
from fweather import get_timeseries
```

```python
stac_url = "https://data.inpe.br/bdc/stac/v1"
```

```python
ts=get_timeseries(
    stac_url=stac_url,
    collection="samet_daily-1",
    start_date="2024-01-01",
    end_date="2024-12-31",
    band="tmean",
    geom=[dict(coordinates=[-11.739, -45.753])],
)
```

```python
ts['values'][:8]
```

```python
import seaborn as sns
import pandas as pd

df = pd.DataFrame(ts)
sns.lineplot(x='timeline', y='values', data=df)
```
